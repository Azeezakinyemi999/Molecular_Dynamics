```python
"""
Cohesive Energy Calculator for Multi-Component Alloys
Using MACE-MP isolated atom reference energies (E0)

Formula:
    E_coh = (sum(N_i * E0_i) - E_bulk) / N_total

"""

import numpy as np

# ── 1. Isolated atom reference energies (eV) from MACE-MP ──────────────────
E0 = {
    'Ni': -1.5178,
    'Al': -0.6941,
    'B':  -0.7036,
    'C':  -1.6918,
    'Cr': -6.2909,
    'Fe': -3.8914,
    'O':  -2.0815,
    'Mo': -4.6289,
}

# ── 2. Alloy composition (atom counts per supercell) ───────────────────────
composition = {
    'Ni': 360,   # 72%  — austenitic FCC matrix
    'Mo': 50,    # 10%  — solid-solution strengthener, corrosion resistance
    'Cr': 42,    #  8.4% — oxidation / corrosion resistance
    'Fe': 18,    #  3.6% — residual from raw materials
    'Al': 10,    #  2%   — deoxidiser, minor gamma-prime former
    'C':  10,    #  2%   — carbide former (M6C, M23C6)
    'B':  10,    #  2%   — grain boundary strengthener, boride former
}

# ── 3. Bulk energies per seed configuration (eV) ──────────────────────────
bulk_energies = {
    7:     -3364.9114,
    42:    -3364.9114,
    111:   -3364.8942,
    1234:  -3368.2721,
    12345: -3365.6745,
}

# ── 4. Compute reference energy sum ───────────────────────────────────────
def reference_energy_sum(composition, E0):
    """Sum of N_i * E0_i over all species."""
    missing = [el for el in composition if el not in E0]
    if missing:
        raise ValueError(f"Missing E0 for elements: {missing}")
    return sum(n * E0[el] for el, n in composition.items())

# ── 5. Cohesive energy for a single configuration ─────────────────────────
def cohesive_energy(E_bulk, composition, E0):
    """Returns cohesive energy in eV/atom (positive = stable)."""
    N_total   = sum(composition.values())
    E_ref_sum = reference_energy_sum(composition, E0)
    return (E_ref_sum - E_bulk) / N_total

# ── 6. Run across all seeds ────────────────────────────────────────────────
def run_all(bulk_energies, composition, E0):
    N_total   = sum(composition.values())
    E_ref_sum = reference_energy_sum(composition, E0)

    print("=" * 55)
    print("  Cohesive Energy Report")
    print("=" * 55)
    print(f"  Composition      : {composition}")
    print(f"  Total atoms      : {N_total}")
    print(f"  Sum(N_i * E0_i)  : {E_ref_sum:.4f} eV")
    print("-" * 55)
    print(f"  {'Seed':<10} {'E_bulk (eV)':<18} {'E_coh (eV/atom)'}")
    print("-" * 55)

    results = {}
    for seed, E_bulk in bulk_energies.items():
        E_coh = cohesive_energy(E_bulk, composition, E0)
        results[seed] = E_coh
        print(f"  {str(seed):<10} {E_bulk:<18.4f} {E_coh:.4f}")

    values = np.array(list(results.values()))
    print("-" * 55)
    print(f"  Mean E_coh       : {values.mean():.4f} eV/atom")
    print(f"  Std dev          : {values.std():.4f} eV/atom")
    print(f"  Min              : {values.min():.4f} eV/atom  (seed {min(results, key=results.get)})")
    print(f"  Max              : {values.max():.4f} eV/atom  (seed {max(results, key=results.get)})")
    print("=" * 55)

    return results

if __name__ == "__main__":
    results = run_all(bulk_energies, composition, E0)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

# ── Data ──────────────────────────────────────────────────────────────────
temperatures = np.array([300, 400, 600, 800])

seeds = {
    '7':     [-3353.7835, -3353.0484, -3336.7312, -3329.5882],
    '42':    [-3356.0868, -3349.8657, -3339.7792, -3329.7885],
    '111':   [-3351.7519, -3349.7031, -3337.9213, -3327.1070],
    '1234':  [-3355.8046, -3351.0733, -3343.8378, -3330.4371],
    '12345': [-3352.0550, -3348.4460, -3340.7002, -3326.0954],
}

data = np.array(list(seeds.values()))   # shape (5, 4)

# ── 1. Descriptive statistics per temperature ──────────────────────────────
mean   = data.mean(axis=0)
std    = data.std(axis=0)
mn     = data.min(axis=0)
mx     = data.max(axis=0)
rng    = mx - mn
cv     = (std / np.abs(mean)) * 100      # coefficient of variation (%)

print("=" * 65)
print("  Descriptive Statistics per Temperature (Hastelloy N Seeds)")
print("=" * 65)
print(f"  {'Stat':<12} {'300 K':>10} {'400 K':>10} {'600 K':>10} {'800 K':>10}")
print("-" * 65)
print(f"  {'Mean (eV)':<12} {mean[0]:>10.4f} {mean[1]:>10.4f} {mean[2]:>10.4f} {mean[3]:>10.4f}")
print(f"  {'Std (eV)':<12} {std[0]:>10.4f} {std[1]:>10.4f} {std[2]:>10.4f} {std[3]:>10.4f}")
print(f"  {'Min (eV)':<12} {mn[0]:>10.4f} {mn[1]:>10.4f} {mn[2]:>10.4f} {mn[3]:>10.4f}")
print(f"  {'Max (eV)':<12} {mx[0]:>10.4f} {mx[1]:>10.4f} {mx[2]:>10.4f} {mx[3]:>10.4f}")
print(f"  {'Range (eV)':<12} {rng[0]:>10.4f} {rng[1]:>10.4f} {rng[2]:>10.4f} {rng[3]:>10.4f}")
print(f"  {'CV (%)':<12} {cv[0]:>10.4f} {cv[1]:>10.4f} {cv[2]:>10.4f} {cv[3]:>10.4f}")
print("=" * 65)

# ── 2. Linear fit of mean energy vs temperature ────────────────────────────
slope, intercept, r, p, se = linregress(temperatures, mean)

print("\n  Linear Fit: Mean Energy vs Temperature")
print("-" * 40)
print(f"  Slope     : {slope:.4f} eV/K")
print(f"  Intercept : {intercept:.4f} eV")
print(f"  R²        : {r**2:.6f}")
print(f"  p-value   : {p:.2e}")

# ── 3. Delta E between consecutive temperatures ────────────────────────────
dT_labels = ['300→400 K', '400→600 K', '600→800 K']
dT_vals   = [400-300, 600-400, 800-600]
delta_mean = np.diff(mean)

print("\n  Mean Energy Change Between Temperatures")
print("-" * 40)
for lbl, dE, dT in zip(dT_labels, delta_mean, dT_vals):
    print(f"  {lbl}  :  ΔE = {dE:.4f} eV  ({dE/dT:.4f} eV/K)")

# ── 4. Outlier check (seeds > 1 std from mean at any T) ───────────────────
print("\n  Outlier Check (> 1 std from mean)")
print("-" * 40)
found = False
for i, (seed, vals) in enumerate(seeds.items()):
    for j, (v, m, s) in enumerate(zip(vals, mean, std)):
        if abs(v - m) > s:
            print(f"  Seed {seed} at {temperatures[j]} K : {v:.4f} eV  "
                  f"(deviation = {abs(v-m):.4f} eV = {abs(v-m)/s:.2f} std)")
            found = True
if not found:
    print("  No outliers detected.")

# ── 5. Plot: mean ± std band + individual seeds + linear fit ───────────────
fig, ax = plt.subplots(figsize=(8, 5))

for seed, vals in seeds.items():
    ax.plot(temperatures, vals, marker='o', linewidth=1, alpha=0.5, label=f'Seed {seed}')

ax.plot(temperatures, mean, 'k-o', linewidth=2.5, label='Mean', zorder=5)
ax.fill_between(temperatures, mean - std, mean + std,
                alpha=0.2, color='black', label='±1 std')

T_fit = np.linspace(300, 800, 200)
ax.plot(T_fit, slope * T_fit + intercept, 'r--', linewidth=1.5,
        label=f'Linear fit (R²={r**2:.4f})')

ax.set_xlabel('Temperature (K)')
ax.set_ylabel('Potential Energy (eV)')
ax.set_title('Hastelloy N — Mean ± Std, Seeds, and Linear Fit')
ax.set_xticks(temperatures)
ax.legend(fontsize=9, loc='lower right')

plt.tight_layout()
plt.savefig('hastelloy_statistics.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ── Raw Data ───────────────────────────────────────────────────────────────
temperatures = [300, 400, 600, 800, 1000, 1200]

D = {
    'Seed 7':     [1.653e-12,  1.964e-12,  1.041e-11,  3.171e-10, 0, 0],
    'Seed 42':    [1.243e-14,  3.325e-11,  9.018e-11,  7.160e-10, 0, 0],
    'Seed 111':   [-2.378e-14, 7.648e-14,  1.867e-10,  1.819e-9, 0, 0 ],
    'Seed 1234':  [-3.800e-15, 1.689e-11,  6.877e-10,  8.324e-10, 0, 0],
    'Seed 12345': [4.443e-14,  6.553e-12,  4.964e-11,  1.466e-9, 0, 0 ],
    'Nickel':     [2.654e-15, -3.231e-14,  6.181e-11,  5.072e-10, 5.751e-09, 1.806e-08 ],
}

R2 = {
    'Seed 7':     [0.9493, 0.9917, 0.9575, 0.9761, 0, 0],
    'Seed 42':    [0.1313, 0.9878, 0.9841, 0.9785, 0, 0],
    'Seed 111':   [0.1288, 0.6464, 0.9803, 0.9901, 0, 0],
    'Seed 1234':  [0.0111, 0.9684, 0.9922, 0.9878, 0, 0],
    'Seed 12345': [0.5136, 0.8456, 0.8949, 0.9897, 0, 0],
    'Nickel':     [0.0057, 0.2440, 0.9385, 0.9896, 0.9849, 0.9320, 0.9934],
}

R2_THRESHOLD = 0.95

# ── Apply Filter ───────────────────────────────────────────────────────────
D_filtered = {}
mask = {}

print("=" * 65)
print("  Reliability Filter  (D > 0  AND  R² >= 0.95)")
print("=" * 65)
print(f"  {'System':<12} {'300 K':>10} {'400 K':>10} {'600 K':>10} {'800 K':>10} {'1000 K':>10} {'1200 K':>10}")
print("-" * 65)

for system in D:
    m = [(d > 0) and (r2 >= R2_THRESHOLD) for d, r2 in zip(D[system], R2[system])]
    mask[system] = m
    D_filtered[system] = [d if k else None for d, k in zip(D[system], m)]
    flags = ['✓' if k else '✗' for k in m]
    print(f"  {system:<12} {flags[0]:>10} {flags[1]:>10} {flags[2]:>10} {flags[3]:>10} {flags[4]:>10} {flags[5]:>10}")

print("=" * 65)
total_pts = sum(len(v) for v in D.values())
kept_pts  = sum(1 for v in D_filtered.values() for x in v if x is not None)
print(f"\n  Total points : {total_pts}")
print(f"  Kept         : {kept_pts}  ({100*kept_pts/total_pts:.0f}%)")
print(f"  Removed      : {total_pts - kept_pts}  ({100*(total_pts-kept_pts)/total_pts:.0f}%)")

# ── Plot ───────────────────────────────────────────────────────────────────
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5), sharey=True)
colors = plt.rcParams['axes.prop_cycle'].by_key()['color']

for idx, (system, values) in enumerate(D.items()):
    col = colors[idx % len(colors)]
    pos_T = [T for T, d in zip(temperatures, values) if d > 0]
    pos_D = [d for d in values if d > 0]
    neg_T = [T for T, d in zip(temperatures, values) if d <= 0]
    neg_D = [abs(d) for d in values if d <= 0]
    if pos_T:
        ax1.semilogy(pos_T, pos_D, 'o-', color=col, linewidth=1.5, label=system)
    if neg_T:
        ax1.semilogy(neg_T, neg_D, 'x', color=col, markersize=9)

    filt_T = [T for T, d in zip(temperatures, D_filtered[system]) if d is not None]
    filt_D = [d for d in D_filtered[system] if d is not None]
    if filt_T:
        ax2.semilogy(filt_T, filt_D, 'o-', color=col, linewidth=1.5, label=system)
    rem_T = [T for T, d in zip(temperatures, D_filtered[system]) if d is None]
    rem_D = [abs(d) for T, d in zip(temperatures, D[system])
             if D_filtered[system][temperatures.index(T)] is None]
    if rem_T:
        ax2.semilogy(rem_T, rem_D, 'x', color=col, markersize=9, alpha=0.25)

for ax, title in zip([ax1, ax2],
                     ['Raw Data (x = negative D)',
                      'Filtered (R² >= 0.95 & D > 0)\n(faded x = removed)']):
    ax.set_xlabel('Temperature (K)')
    ax.set_xticks(temperatures)
    ax.grid(True, which='both', linestyle='--', alpha=0.4)
    ax.set_title(title, fontsize=10)

ax1.set_ylabel('D (m²/s)')
ax1.legend(fontsize=8, loc='upper left')
ax2.legend(fontsize=8, loc='upper left')
fig.suptitle('Step 1 — Reliability Filtering', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('step1_reliability_filter.png', dpi=150, bbox_inches='tight')
plt.show()
```

#### Step 3

```python
# ── Arrhenius Analysis ─────────────────────────────────────────────────────
# Paste D_filtered and temperatures from Step 1, then run this

from scipy.stats import linregress

kB = 8.617333e-5   # eV/K

print("=" * 60)
print("  Arrhenius Analysis  ln(D) vs 1/T")
print("=" * 60)
print(f"  {'System':<12} {'Ea (eV)':>10} {'D0 (m²/s)':>14} {'R²':>8}")
print("-" * 60)

all_systems = hastelloy_systems + ['Nickel']
Ea_vals, D0_vals = [], []

fig, ax = plt.subplots(figsize=(8, 5))
colors = plt.rcParams['axes.prop_cycle'].by_key()['color']

for idx, system in enumerate(all_systems):
    T_arr = [T for T, d in zip(temperatures, D_filtered[system]) if d is not None]
    D_arr = [d for d in D_filtered[system] if d is not None]
    if len(T_arr) < 2:
        print(f"  {system:<12}  insufficient data points")
        continue

    inv_T  = [1 / T for T in T_arr]
    ln_D   = [np.log(d) for d in D_arr]
    slope, intercept, r, p, _ = linregress(inv_T, ln_D)

    Ea  = -slope * kB          # eV
    D0  = np.exp(intercept)    # m²/s
    Ea_vals.append(Ea)
    D0_vals.append(D0)

    print(f"  {system:<12} {Ea:>10.4f} {D0:>14.4e} {r**2:>8.4f}")

    col = colors[idx % len(colors)]
    ls  = '--' if system == 'Nickel' else '-'
    ax.plot(inv_T, ln_D, 'o', color=col, markersize=6)
    x_fit = np.linspace(min(inv_T), max(inv_T), 100)
    ax.plot(x_fit, slope * x_fit + intercept, ls, color=col,
            linewidth=1.5, label=f'{system} (Ea={Ea:.3f} eV)')

print("=" * 60)
hn_Ea = [e for s, e in zip(all_systems, Ea_vals) if s != 'Nickel']
if hn_Ea:
    print(f"\n  Hastelloy N mean Ea : {np.mean(hn_Ea):.4f} ± {np.std(hn_Ea):.4f} eV")

ax.set_xlabel('1/T  (K⁻¹)')
ax.set_ylabel('ln(D)  [D in m²/s]')
ax.set_title('Step 3 — Arrhenius Plot: ln(D) vs 1/T')
ax.legend(fontsize=8, loc='upper right')
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig('step3_arrhenius.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
"""
Surface Relaxation — Extended Analysis Plots
=============================================
Plot A: Exponential decay fit — Δz/d₁₁₁ vs layer per seed + decay constants
Plot B: Seed variance (std) of Δz/d₁₁₁ vs layer depth
Plot C: z_after vs layer number (reference — actual plane positions)

Fit model (measured from surface downward):
    Δz/d₁₁₁(depth) = A · exp(−depth / ξ)
    depth = layer_surface − layer   (so surface = depth 0)
    ξ = characteristic penetration depth [layer units]
"""

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from scipy.optimize import curve_fit

# ── Data ──────────────────────────────────────────────────────────────────────

layers    = np.arange(1, 13)
d_ideal   = 2.0327   # Å

z_before = np.array([
    15.0000, 17.0327, 19.0655, 21.0982,
    23.1309, 25.1637, 27.1964, 29.2291,
    31.2619, 33.2946, 35.3273, 37.3601,
])

z_after = {
    "Seed 7":     np.array([15.0000, 17.0327, 19.0655, 21.0982,
                             23.2182, 25.3253, 27.4795, 29.5854,
                             31.6238, 33.7358, 35.7586, 37.7472]),
    "Seed 42":    np.array([15.0000, 17.0327, 19.0655, 21.0982,
                             23.2739, 25.2474, 27.3691, 29.4717,
                             31.5379, 33.6666, 35.7775, 37.8874]),
    "Seed 111":   np.array([15.0000, 17.0327, 19.0655, 21.0982,
                             23.2437, 25.2646, 27.4014, 29.4252,
                             31.5245, 33.6954, 35.8990, 37.9319]),
    "Seed 1234":  np.array([15.0000, 17.0327, 19.0655, 21.0982,
                             23.2095, 25.3139, 27.3955, 29.4977,
                             31.5831, 33.6733, 35.7647, 37.7873]),
    "Seed 12345": np.array([15.0000, 17.0327, 19.0655, 21.0982,
                             23.2329, 25.3259, 27.4209, 29.4922,
                             31.6028, 33.6935, 35.7949, 37.8953]),
}

dz_pct = {
    "Seed 7":     [0.00, 0.00, 0.00, 0.00,  4.29,  7.95, 13.93, 17.53, 17.81, 21.70, 21.21, 19.05],
    "Seed 42":    [0.00, 0.00, 0.00, 0.00,  7.03,  4.12,  8.50, 11.93, 13.58, 18.30, 22.15, 25.94],
    "Seed 111":   [0.00, 0.00, 0.00, 0.00,  5.55,  4.97, 10.08,  9.65, 12.92, 19.72, 28.12, 28.13],
    "Seed 1234":  [0.00, 0.00, 0.00, 0.00,  3.86,  7.39,  9.80, 13.21, 15.80, 18.63, 21.51, 21.02],
    "Seed 12345": [0.00, 0.00, 0.00, 0.00,  5.01,  7.98, 11.04, 12.94, 16.77, 19.62, 23.00, 26.33],
}

seed_labels = list(dz_pct.keys())
matrix      = np.array([dz_pct[s] for s in seed_labels])   # (5, 12)

seed_colors = {
    "Seed 7":     "#4E9AF1",
    "Seed 42":    "#F4A261",
    "Seed 111":   "#2EC4B6",
    "Seed 1234":  "#E94F37",
    "Seed 12345": "#A57BCA",
}

# relaxed layers only (indices 4-11, i.e. layers 5-12)
relaxed_idx    = np.arange(4, 12)
relaxed_layers = layers[relaxed_idx]          # [5..12]
layer_surface  = 12

# depth from surface: surface = 0, layer 5 = depth 7
depth = layer_surface - relaxed_layers        # [7, 6, 5, 4, 3, 2, 1, 0]

# ── Exponential model ─────────────────────────────────────────────────────────
def exp_decay(d, A, xi):
    """Δz/d₁₁₁ = A · exp(−d / ξ)"""
    return A * np.exp(-d / xi)

# ── Figure ────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(
    1, 3,
    figsize=(18, 6),
    facecolor="white",
)
fig.subplots_adjust(wspace=0.38)

# ─────────────────────────────────────────────────────────────────────────────
# PLOT A — Exponential decay fit
# ─────────────────────────────────────────────────────────────────────────────
ax = axes[0]
ax.set_facecolor("white")

fit_results = {}

for seed in seed_labels:
    y_relax = np.array(dz_pct[seed])[relaxed_idx]
    color   = seed_colors[seed]

    # raw data points
    ax.plot(
        relaxed_layers, y_relax,
        color=color, linewidth=0, marker="o",
        markersize=6, markeredgecolor="white", markeredgewidth=0.6,
        label=seed, zorder=4,
    )

    # fit — use depth (surface=0) so A is the surface amplitude
    try:
        popt, pcov = curve_fit(
            exp_decay, depth, y_relax,
            p0=[25, 3.0],
            bounds=([0, 0.1], [100, 20]),
            maxfev=5000,
        )
        A_fit, xi_fit = popt
        fit_results[seed] = {"A": A_fit, "xi": xi_fit}

        # smooth fit line over depth range 0→7, convert back to layer
        d_smooth     = np.linspace(0, 7.5, 200)
        layer_smooth = layer_surface - d_smooth
        y_smooth     = exp_decay(d_smooth, A_fit, xi_fit)

        ax.plot(
            layer_smooth, y_smooth,
            color=color, linewidth=1.5, linestyle="--", alpha=0.75, zorder=3,
        )
    except RuntimeError:
        fit_results[seed] = {"A": np.nan, "xi": np.nan}

# frozen/relaxed boundary
ax.axvline(x=4.5, color="#888888", linewidth=0.8, linestyle=":", alpha=0.7)
ax.text(4.55, 29.5, "relaxed →", color="#555555", fontsize=8, va="top")
ax.text(4.45, 29.5, "← frozen",  color="#555555", fontsize=8, va="top", ha="right")
ax.axhline(y=0, color="#CCCCCC", linewidth=0.8)

# annotate decay constants as a table inside the axes
xi_text = "ξ (layers)\n" + "─" * 16 + "\n"
for seed, res in fit_results.items():
    label = seed.replace("Seed ", "S")
    xi_text += f"{label:<8}  {res['xi']:.2f}\n"
ax.text(
    0.03, 0.97, xi_text.strip(),
    transform=ax.transAxes,
    fontsize=7.5, va="top", ha="left",
    fontfamily="monospace",
    color="#1A1A1A",
    bbox=dict(boxstyle="round,pad=0.4", facecolor="#F5F5F5",
              edgecolor="#CCCCCC", alpha=0.9),
)

ax.set_xlabel("Layer number (bottom → top)", color="#1A1A1A", fontsize=11)
ax.set_ylabel("Δz / d₁₁₁ (%)", color="#1A1A1A", fontsize=11)
ax.set_title("(A)  Relaxation Depth — Exponential Decay Fit\n"
             r"$\Delta z/d_{111} = A \cdot e^{-\mathrm{depth}/\xi}$",
             color="#1A1A1A", fontsize=11, pad=8)
ax.set_xticks(layers)
ax.set_xticklabels(
    [str(l) if l != 12 else "12\n(surf)" for l in layers],
    color="#1A1A1A", fontsize=8.5,
)
ax.tick_params(axis="y", colors="#1A1A1A", labelsize=9)
ax.set_xlim(0.5, 12.8)
ax.set_ylim(-2, 33)
ax.grid(which="major", color="#E8E8E8", linewidth=0.6)
ax.grid(which="minor", color="#F4F4F4", linewidth=0.3)
ax.yaxis.set_minor_locator(ticker.MultipleLocator(2))
for spine in ax.spines.values():
    spine.set_edgecolor("#CCCCCC")
ax.legend(fontsize=8, framealpha=0.9, facecolor="white",
          edgecolor="#CCCCCC", labelcolor="#1A1A1A", loc="upper left",
          bbox_to_anchor=(0.01, 0.72))

# ─────────────────────────────────────────────────────────────────────────────
# PLOT B — Seed variance (std) vs layer
# ─────────────────────────────────────────────────────────────────────────────
ax2 = axes[1]
ax2.set_facecolor("white")

std_all   = matrix.std(axis=0, ddof=1)      # all 12 layers
mean_all  = matrix.mean(axis=0)

# bar chart coloured by magnitude
bar_colors = plt.cm.Oranges(std_all / std_all.max() * 0.85 + 0.1)
bars = ax2.bar(layers, std_all, color=bar_colors, edgecolor="white",
               linewidth=0.6, zorder=3)

# overlay mean profile (secondary visual reference, right y-axis)
ax2b = ax2.twinx()
ax2b.plot(layers, mean_all, color="#4E9AF1", linewidth=1.6,
          linestyle="-", marker="o", markersize=4,
          markerfacecolor="#4E9AF1", markeredgecolor="white",
          markeredgewidth=0.5, alpha=0.7, label="Mean Δz/d₁₁₁",
          zorder=5)
ax2b.set_ylabel("Mean Δz / d₁₁₁ (%)", color="#4E9AF1", fontsize=10)
ax2b.tick_params(axis="y", colors="#4E9AF1", labelsize=8)
ax2b.set_ylim(-2, 38)
ax2b.spines["right"].set_edgecolor("#4E9AF1")

# frozen/relaxed boundary
ax2.axvline(x=4.5, color="#888888", linewidth=0.8, linestyle=":", alpha=0.7)
ax2.text(4.55, std_all.max() * 0.96, "relaxed →",
         color="#555555", fontsize=8, va="top")
ax2.text(4.45, std_all.max() * 0.96, "← frozen",
         color="#555555", fontsize=8, va="top", ha="right")

# annotate peak variance layer
peak_layer = layers[np.argmax(std_all)]
ax2.annotate(
    f"peak σ\nlayer {peak_layer}",
    xy=(peak_layer, std_all.max()),
    xytext=(peak_layer - 1.8, std_all.max() * 0.88),
    fontsize=7.5, color="#C0392B",
    arrowprops=dict(arrowstyle="->", color="#C0392B", lw=1.0),
)

ax2.set_xlabel("Layer number (bottom → top)", color="#1A1A1A", fontsize=11)
ax2.set_ylabel("σ  (Δz / d₁₁₁)  across seeds  (%)", color="#1A1A1A", fontsize=11)
ax2.set_title("(B)  Seed Variance vs Depth\nσ of Δz/d₁₁₁ across all seeds",
              color="#1A1A1A", fontsize=11, pad=8)
ax2.set_xticks(layers)
ax2.set_xticklabels(
    [str(l) if l != 12 else "12\n(surf)" for l in layers],
    color="#1A1A1A", fontsize=8.5,
)
ax2.tick_params(axis="y", colors="#1A1A1A", labelsize=9)
ax2.set_xlim(0.5, 12.5)
ax2.set_ylim(0, std_all.max() * 1.25)
ax2.grid(axis="y", color="#E8E8E8", linewidth=0.6, zorder=0)
for spine in ax2.spines.values():
    spine.set_edgecolor("#CCCCCC")
ax2b.spines["top"].set_edgecolor("#CCCCCC")
ax2b.spines["bottom"].set_edgecolor("#CCCCCC")
ax2b.spines["left"].set_edgecolor("#CCCCCC")

# combined legend
lines2, labels2 = ax2b.get_legend_handles_labels()
from matplotlib.patches import Patch
patch = Patch(facecolor="#F0A868", edgecolor="white", label="σ (bar)")
ax2.legend(handles=[patch] + lines2, labels=["σ (bar)"] + labels2,
           fontsize=8, framealpha=0.9, facecolor="white",
           edgecolor="#CCCCCC", labelcolor="#1A1A1A", loc="upper left")

# ─────────────────────────────────────────────────────────────────────────────
# PLOT C — z_after vs Layer (reference)
# ─────────────────────────────────────────────────────────────────────────────
ax3 = axes[2]
ax3.set_facecolor("white")

# ideal reference line
z_ideal = z_before
ax3.plot(layers, z_ideal, color="#AAAAAA", linewidth=1.2,
         linestyle="--", label="Ideal (unrelaxed)", zorder=2)

for seed in seed_labels:
    ax3.plot(
        layers, z_after[seed],
        color=seed_colors[seed], linewidth=1.5,
        marker="o", markersize=4,
        markerfacecolor=seed_colors[seed],
        markeredgecolor="white", markeredgewidth=0.5,
        label=seed, zorder=3,
    )

# shade the deviation region (max envelope)
z_mat    = np.array([z_after[s] for s in seed_labels])
z_min    = z_mat.min(axis=0)
z_max    = z_mat.max(axis=0)
ax3.fill_between(layers, z_min, z_max,
                 color="#AAAAAA", alpha=0.12, label="Seed envelope", zorder=1)

# frozen/relaxed boundary
ax3.axvline(x=4.5, color="#888888", linewidth=0.8, linestyle=":", alpha=0.7)
ax3.text(4.55, 16.5, "relaxed →", color="#555555", fontsize=8, va="bottom")
ax3.text(4.45, 16.5, "← frozen",  color="#555555", fontsize=8, va="bottom", ha="right")

ax3.set_xlabel("Layer number (bottom → top)", color="#1A1A1A", fontsize=11)
ax3.set_ylabel("z_after (Å)", color="#1A1A1A", fontsize=11)
ax3.set_title("(C)  Actual Plane Positions Post-Relaxation\n"
              "Kinks = local compression / expansion zones",
              color="#1A1A1A", fontsize=11, pad=8)
ax3.set_xticks(layers)
ax3.set_xticklabels(
    [str(l) if l != 12 else "12\n(surf)" for l in layers],
    color="#1A1A1A", fontsize=8.5,
)
ax3.tick_params(axis="y", colors="#1A1A1A", labelsize=9)
ax3.set_xlim(0.5, 12.5)
ax3.grid(which="major", color="#E8E8E8", linewidth=0.6)
for spine in ax3.spines.values():
    spine.set_edgecolor("#CCCCCC")
ax3.legend(fontsize=8, framealpha=0.9, facecolor="white",
           edgecolor="#CCCCCC", labelcolor="#1A1A1A",
           loc="upper left", ncol=1)

# ── Super title ───────────────────────────────────────────────────────────────
fig.suptitle(
    "FCC (111) Surface Relaxation — Extended Analysis  |  "
    "Hastelloy X/N  |  d₁₁₁ = 2.0327 Å  |  MACE-MP-0 potential",
    color="#1A1A1A", fontsize=11, y=1.01,
)

plt.tight_layout()
plt.savefig(
    "surface_relaxation_extended.png",
    dpi=200,
    bbox_inches="tight",
    facecolor="white",
)
print("Saved: surface_relaxation_extended.png")

# ── Print decay constants table ───────────────────────────────────────────────
print("\nExponential Decay Fit Results")
print("=" * 40)
print(f"{'Seed':<12}  {'A (%)':>8}  {'ξ (layers)':>12}")
print("-" * 40)
for seed, res in fit_results.items():
    print(f"{seed:<12}  {res['A']:>8.2f}  {res['xi']:>12.3f}")
print("=" * 40)
print("Model: Δz/d₁₁₁ = A · exp(−depth/ξ),  depth = 12 − layer")

plt.show()
```

```python
"""
Interlayer Spacing Analysis — Plots 1, 6, 7
============================================
Panel 1: Δ% vs spacing index — one line per seed + mean ± σ band
Panel 6: Heatmap of Δ% (spacing × seed)
Panel 7: Actual d(n,n+1) in Å vs spacing index with ideal reference line

Seeds: 7, 42, 111, 1234, 12345
d_ideal = 2.0327 Å  (FCC 111)
"""

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from matplotlib.colors import TwoSlopeNorm
from matplotlib.lines import Line2D
import matplotlib.patches as mpatches

# ── Data ──────────────────────────────────────────────────────────────────────

d_ideal = 2.0327  # Å

spacing_labels = [
    "d(1,2)", "d(2,3)", "d(3,4)",
    "d(4,5)", "d(5,6)", "d(6,7)", "d(7,8)",
    "d(8,9)", "d(9,10)", "d(10,11)", "d(11,12)",
]
n_spacings = len(spacing_labels)
x = np.arange(n_spacings)

# Δ% values per seed
delta_pct = {
    "Seed 7":     [ 0.00, -0.00,  0.00,  4.29,  3.66,  5.97,  3.60,  0.28,  3.90, -0.49, -2.17],
    "Seed 42":    [ 0.00, -0.00,  0.00,  7.03, -2.91,  4.37,  3.44,  1.65,  4.72,  3.85,  3.80],
    "Seed 111":   [ 0.00, -0.00,  0.00,  5.55, -0.58,  5.12, -0.44,  3.27,  6.80,  8.41,  0.01],
    "Seed 1234":  [ 0.00, -0.00,  0.00,  3.86,  3.52,  2.41,  3.41,  2.59,  2.82,  2.89, -0.50],
    "Seed 12345": [ 0.00, -0.00,  0.00,  5.01,  2.97,  3.06,  1.90,  3.83,  2.85,  3.38,  3.33],
}

# Actual d values (Å) per seed
d_vals = {
    "Seed 7":     [2.0327, 2.0327, 2.0327, 2.1200, 2.1071, 2.1542, 2.1059, 2.0384, 2.1120, 2.0228, 1.9887],
    "Seed 42":    [2.0327, 2.0327, 2.0327, 2.1757, 1.9735, 2.1217, 2.1026, 2.0662, 2.1287, 2.1109, 2.1099],
    "Seed 111":   [2.0327, 2.0327, 2.0327, 2.1455, 2.0209, 2.1367, 2.0239, 2.0993, 2.1709, 2.2036, 2.0329],
    "Seed 1234":  [2.0327, 2.0327, 2.0327, 2.1113, 2.1044, 2.0817, 2.1021, 2.0855, 2.0901, 2.0914, 2.0226],
    "Seed 12345": [2.0327, 2.0327, 2.0327, 2.1347, 2.0930, 2.0949, 2.0714, 2.1105, 2.0907, 2.1014, 2.1004],
}

seed_labels = list(delta_pct.keys())
matrix_pct  = np.array([delta_pct[s] for s in seed_labels])   # (5, 11)
mean_pct    = matrix_pct.mean(axis=0)
std_pct     = matrix_pct.std(axis=0, ddof=1)

seed_colors = {
    "Seed 7":     "#4E9AF1",
    "Seed 42":    "#F4A261",
    "Seed 111":   "#2EC4B6",
    "Seed 1234":  "#E94F37",
    "Seed 12345": "#A57BCA",
}

frozen_boundary = 2.5   # between d(3,4) and d(4,5)

# ── Figure ────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(
    1, 3,
    figsize=(19, 6.5),
    facecolor="white",
    gridspec_kw={"width_ratios": [1.6, 1, 1.4]},
)
fig.subplots_adjust(wspace=0.40)

# ══════════════════════════════════════════════════════════════════════════════
# PANEL 1 — Δ% line profiles + mean ± σ band
# ══════════════════════════════════════════════════════════════════════════════
ax1 = axes[0]
ax1.set_facecolor("white")

# frozen zone background
ax1.axvspan(-0.5, frozen_boundary, color="#F5F5F5", alpha=0.8, zorder=0)

# ±σ band
ax1.fill_between(
    x, mean_pct - std_pct, mean_pct + std_pct,
    color="#4E9AF1", alpha=0.13, label="±1σ band", zorder=2,
)

# individual seed lines
for seed in seed_labels:
    ax1.plot(
        x, delta_pct[seed],
        color=seed_colors[seed], linewidth=1.4,
        linestyle="--", marker="o", markersize=5,
        markeredgecolor="white", markeredgewidth=0.5,
        alpha=0.65, label=seed, zorder=3,
    )

# mean line
ax1.plot(
    x, mean_pct,
    color="#1A1A1A", linewidth=2.2,
    marker="o", markersize=5,
    markerfacecolor="#333333", markeredgecolor="white",
    markeredgewidth=0.6, label="Mean", zorder=5,
)

# zero reference
ax1.axhline(0, color="#AAAAAA", linewidth=0.9, linestyle="-", zorder=1)

# frozen/relaxed boundary
ax1.axvline(frozen_boundary, color="#888888", linewidth=0.8,
            linestyle=":", alpha=0.8)
ax1.text(frozen_boundary + 0.08,  8.6, "relaxed →",
         color="#555555", fontsize=8, va="top")
ax1.text(frozen_boundary - 0.08,  8.6, "← frozen",
         color="#555555", fontsize=8, va="top", ha="right")

# highlight contraction zones (negative mean) with subtle red tint
for i, val in enumerate(mean_pct):
    if val < 0:
        ax1.axvspan(i - 0.45, i + 0.45,
                    color="#E94F37", alpha=0.07, zorder=0)

ax1.set_xticks(x)
ax1.set_xticklabels(spacing_labels, rotation=40, ha="right",
                    color="#1A1A1A", fontsize=8.5)
ax1.set_ylabel("Δ%  vs ideal d₁₁₁", color="#1A1A1A", fontsize=11)
ax1.set_xlabel("Interlayer spacing", color="#1A1A1A", fontsize=11)
ax1.set_title("(1)  Δ% Profile — All Seeds + Mean ± σ",
              color="#1A1A1A", fontsize=11, pad=10)
ax1.set_xlim(-0.5, n_spacings - 0.5)
ax1.set_ylim(-5.5, 11.5)
ax1.yaxis.set_minor_locator(ticker.MultipleLocator(1))
ax1.grid(which="major", color="#E8E8E8", linewidth=0.6, zorder=0)
ax1.grid(which="minor", color="#F4F4F4", linewidth=0.3, zorder=0)
for spine in ax1.spines.values():
    spine.set_edgecolor("#CCCCCC")

legend_handles = [
    Line2D([0], [0], color=seed_colors[s], linewidth=1.4,
           linestyle="--", marker="o", markersize=4,
           markeredgecolor="white", label=s)
    for s in seed_labels
] + [
    Line2D([0], [0], color="#1A1A1A", linewidth=2.2,
           marker="o", markersize=4,
           markeredgecolor="white", label="Mean"),
    mpatches.Patch(facecolor="#4E9AF1", alpha=0.3,
                   edgecolor="none", label="±1σ band"),
]
ax1.legend(handles=legend_handles, fontsize=7.8, framealpha=0.9,
           facecolor="white", edgecolor="#CCCCCC",
           labelcolor="#1A1A1A", loc="upper right", ncol=2)

# ══════════════════════════════════════════════════════════════════════════════
# PANEL 6 — Heatmap: Δ% (spacing × seed)
# ══════════════════════════════════════════════════════════════════════════════
ax6 = axes[1]
ax6.set_facecolor("white")

vmax = np.abs(matrix_pct).max() + 0.5
norm = TwoSlopeNorm(vmin=-vmax, vcenter=0, vmax=vmax)

im = ax6.imshow(
    matrix_pct.T,          # (11 spacings, 5 seeds), origin="upper" → d(1,2) at top
    aspect="auto",
    cmap="RdYlBu_r",
    norm=norm,
    origin="upper",
    interpolation="nearest",
)

# cell gridlines
for gx in np.arange(-0.5, 5, 1):
    ax6.axvline(gx, color="white", linewidth=0.8)
for gy in np.arange(-0.5, 11, 1):
    ax6.axhline(gy, color="white", linewidth=0.8)

# frozen/relaxed boundary (rows 0-2 frozen, row 3 onward relaxed)
ax6.axhline(2.5, color="#333333", linewidth=1.3, linestyle="--", alpha=0.85)
ax6.text(4.48, 2.42, "relaxed ↓", color="#333333",
         fontsize=7, va="bottom", ha="right")

# annotate cells
for i in range(len(seed_labels)):
    for j in range(n_spacings):
        val = matrix_pct[i, j]
        tc  = "white" if (abs(val) > 5.0 or val < -1.5) else "#111111"
        ax6.text(i, j, f"{val:+.1f}",
                 ha="center", va="center",
                 fontsize=6.5, color=tc, fontweight="bold")

ax6.set_xticks(range(len(seed_labels)))
ax6.set_xticklabels(
    [s.replace("Seed ", "S") for s in seed_labels],
    color="#1A1A1A", fontsize=9,
)
ax6.set_yticks(range(n_spacings))
ax6.set_yticklabels(spacing_labels, color="#1A1A1A", fontsize=8)
ax6.set_xlabel("Seed", color="#1A1A1A", fontsize=11)
ax6.set_title("(6)  Δ% Heatmap\n(spacing × seed)",
              color="#1A1A1A", fontsize=11, pad=10)
for spine in ax6.spines.values():
    spine.set_edgecolor("#CCCCCC")

cbar = fig.colorbar(im, ax=ax6, fraction=0.046, pad=0.04)
cbar.set_label("Δ%  vs ideal d₁₁₁", color="#1A1A1A", fontsize=9)
cbar.ax.yaxis.set_tick_params(color="#1A1A1A", labelsize=8)
plt.setp(cbar.ax.yaxis.get_ticklabels(), color="#1A1A1A")
cbar.outline.set_edgecolor("#CCCCCC")

# ══════════════════════════════════════════════════════════════════════════════
# PANEL 7 — Actual d (Å) vs spacing index with ideal reference
# ══════════════════════════════════════════════════════════════════════════════
ax7 = axes[2]
ax7.set_facecolor("white")

# frozen zone background
ax7.axvspan(-0.5, frozen_boundary, color="#F5F5F5", alpha=0.8, zorder=0)

# ideal reference
ax7.axhline(d_ideal, color="#888888", linewidth=1.3,
            linestyle="--", label=f"Ideal  {d_ideal} Å", zorder=2)

# seed lines
for seed in seed_labels:
    ax7.plot(
        x, d_vals[seed],
        color=seed_colors[seed], linewidth=1.5,
        marker="o", markersize=5,
        markeredgecolor="white", markeredgewidth=0.5,
        label=seed, zorder=3, alpha=0.85,
    )

# seed envelope shading
d_mat = np.array([d_vals[s] for s in seed_labels])
ax7.fill_between(x, d_mat.min(axis=0), d_mat.max(axis=0),
                 color="#AAAAAA", alpha=0.13,
                 label="Seed envelope", zorder=1)

# annotate notable outliers (|Δ%| > 5.5)
annotated = set()
for seed in seed_labels:
    for i in range(n_spacings):
        pct  = delta_pct[seed][i]
        yval = d_vals[seed][i]
        key  = (i, seed)
        if abs(pct) > 5.5 and key not in annotated:
            annotated.add(key)
            offset = 0.007 if yval > d_ideal else -0.010
            ax7.annotate(
                f"{pct:+.1f}%",
                xy=(i, yval),
                xytext=(i + 0.18, yval + offset),
                fontsize=6.5, color=seed_colors[seed],
                arrowprops=dict(arrowstyle="->",
                                color=seed_colors[seed], lw=0.7),
            )

# frozen/relaxed boundary
ax7.axvline(frozen_boundary, color="#888888", linewidth=0.8,
            linestyle=":", alpha=0.8)
ax7.text(frozen_boundary + 0.08, 2.215, "relaxed →",
         color="#555555", fontsize=8, va="top")
ax7.text(frozen_boundary - 0.08, 2.215, "← frozen",
         color="#555555", fontsize=8, va="top", ha="right")

ax7.set_xticks(x)
ax7.set_xticklabels(spacing_labels, rotation=40, ha="right",
                    color="#1A1A1A", fontsize=8.5)
ax7.set_ylabel("d(n, n+1)  (Å)", color="#1A1A1A", fontsize=11)
ax7.set_xlabel("Interlayer spacing", color="#1A1A1A", fontsize=11)
ax7.set_title("(7)  Actual Interlayer Spacings Post-Relaxation\n"
              "Kinks = local compression / expansion zones",
              color="#1A1A1A", fontsize=11, pad=10)
ax7.set_xlim(-0.5, n_spacings - 0.5)
ax7.set_ylim(1.95, 2.24)
ax7.yaxis.set_major_formatter(ticker.FormatStrFormatter("%.4f"))
ax7.yaxis.set_minor_locator(ticker.MultipleLocator(0.005))
ax7.grid(which="major", color="#E8E8E8", linewidth=0.6)
ax7.grid(which="minor", color="#F4F4F4", linewidth=0.3)
for spine in ax7.spines.values():
    spine.set_edgecolor("#CCCCCC")
ax7.legend(fontsize=7.8, framealpha=0.9, facecolor="white",
           edgecolor="#CCCCCC", labelcolor="#1A1A1A",
           loc="upper right", ncol=1)

# ── Super title ───────────────────────────────────────────────────────────────
fig.suptitle(
    "FCC (111) Interlayer Spacing Analysis  |  "
    "Hastelloy X/N  |  d₁₁₁ = 2.0327 Å  |  MACE-MP-0 potential",
    color="#1A1A1A", fontsize=11, y=1.01,
)

plt.tight_layout()
plt.savefig(
    "interlayer_spacing_plots.png",
    dpi=200,
    bbox_inches="tight",
    facecolor="white",
)
print("Saved: interlayer_spacing_plots.png")
plt.show()
```
