```python
import ase
```

```python
from ase.io import read, write
from ase.visualize import view
```

```python
from ase.build import fcc111
from ase.visualize import view
slab = fcc111('Ni', size=(5, 6, 12), vacuum=15.0, periodic=True)
```

```python
view(slab, viewer='x3d')
```

```python
from ase.io import write, read
write('slab_Ni_5x6x12.xyz', slab, format='xyz')
```

```python
write('slab_Ni_5x6x12.lammps', slab, format='lammps-data')
```

```python
write('slab_Ni_5x6x12.vasp', slab, format='vasp')
```
